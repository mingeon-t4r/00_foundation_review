```python
# 1. 변수와 자료형
customer_name = "Kim"
sales = 125000
growth_rate = 0.083
is_active = True
cancelled_at = None
```

각 변수의 자료형과 용도를 설명
customer_name : str 문자열
sales : int 정수
growth_rate : float 실수
is_active : bool 논리
cancelled_at : none 빈값

```python
# 2. 문자열과 f-string
print(f"고객 : {customer_name}")
print(f"매출 : {sales:,}원")
print(f"성장률 : {growth_rate:.1%}")
```

:, 는 천단위 표시
:.1% 는 소수를 백분율로 바꾸고 소수점 한 자리 표시
:.0f 는 소수점 없이 실수 표시

```python
# 3. 리스트-튜플-딕셔너리-집합
customer = {
    "customer_id" : 1,
    "name" : "Kim",
    "region" : "Seoul"
}

discount = customer.get("discount", 0)
print(discount)
```

리스트(list) : 순서가 있고 변경 가능한 자료형	[1, 2, 3], ["A", "B"]
튜플(tuple) : 순서가 있고 변경 불가능한 자료형	(1, 2, 3)
딕셔너리(dict) : 키-값 쌍의 집합	{"name": "Kim", "age": 20}

```python
# 4. 조건문
sales = 180000

if sales >= 200000 :
    grade = "VIP"
elif sales >= 100000 :
    grade = "Regular"
else:
    grade = "Basic"

print(grade)
```

```python
# 5. 반복문
for i in range(1,20):
    if i % 2 == 0:
        print(i)

sales = [8000, 12000, 15000]
for i, sale in enumerate(sales):
    if sale >= 10000:
        print(i,sale)

customers = ["Kim", "Park", "Choi"]
for customer, sale in zip(customers, sales):
    print(f"{customer}의 판매금액은 {sale:,}원입니다")

scores = {
    "국어" : 80,
    "영어" : 90,
    "수학" : 98,
    "과학" : 76
}
for subject, score in scores.items():
    print(f"{subject} : {score}점")
```

```python
# 6. 함수
def calculate_discount(price, discount_rate=0):
    discount_amount = price * discount_rate
    final_price = price - discount_amount
    return discount_amount, final_price
```

매개변수는 무엇인가 : price, discount_rate
기본값은 무엇인가 : discount_rate = 0
반환값은 무엇인가 : discount_amount, final_price
함수가 필요한 이유는 무엇인가 : 코드 반복없이 계산 방식을 정해서 하나로 관리
