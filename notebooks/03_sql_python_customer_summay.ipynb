```python
from pathlib import Path
import sqlite3
import pandas as pd

# 데이터베이스 경로 확인
db_path = Path("D:/Study/SQL/database/retail_lab.db").resolve()

print("DB 경로:", db_path)
print("DB 파일 존재 여부:", db_path.exists())

query = """
SELECT
    region,
    COUNT(*) AS customer_count
FROM customers
GROUP BY region
ORDER BY customer_count DESC;
"""

print("실행할 SQL:")
print(query)

with sqlite3.connect(db_path) as conn:
    df = pd.read_sql_query(query, conn)

print(df)
```

```python
df.plot.bar(
    x = "region",
    y = "customer_count"
)

import matplotlib.pyplot as plt

plt.savefig("../outputs/region_customer_count.png", dpi = 150, bbox_inches = "tight")
```
